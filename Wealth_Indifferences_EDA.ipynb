```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# Load the dataset
data = pd.read_csv("/content/net worth by quarter.csv", encoding='ISO-8859-1')
data.head(10)
```

```python
data.rename(columns={'WFRBL99T999219': '99.9% to 100%'}, inplace=True)
data.rename(columns={'WFRBLB50107': '99% to 99.9%'}, inplace=True)
data.rename(columns={'WFRBLN09053': '99% to 100%'}, inplace=True)
data.rename(columns={'WFRBLN40080': '90% to 99%'}, inplace=True)
data.rename(columns={'WFRBLT01026': '50% to 90%'}, inplace=True)
data.rename(columns={'WFRBLTP1246': '0% to 50%'}, inplace=True)
```

```python
data.head(10)
```

```python
data['observation_date'] = pd.to_datetime(data['observation_date'])
filtered_data = data[(data['observation_date'].dt.year >= 2004) & (data['observation_date'].dt.year <= 2024)]
```

```python
plt.figure(figsize=(12, 6))
plt.plot(filtered_data['observation_date'], filtered_data['99.9% to 100%'], label='99.9% to 100%')
plt.plot(filtered_data['observation_date'], filtered_data['99% to 99.9%'], label='99% to 99.9%')
plt.plot(filtered_data['observation_date'], filtered_data['90% to 99%'], label='90% to 99%')
plt.plot(filtered_data['observation_date'], filtered_data['0% to 50%'], label='Bottom 50%')
plt.title('Wealth Distribution Over Time (2004-2024)')
plt.legend(loc = "upper left", bbox_to_anchor = (1.0, 1.0))
plt.xlabel('Observation Date')
plt.ylabel('Wealth (in millions)')
plt.show()
```
